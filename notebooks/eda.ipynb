```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Adjust path if your notebook root is notebooks/
df = pd.read_csv('../data/grid_layout_10x10.csv', index_col=0)
df
```

```python
counts = df.values.flatten()
pd.Series(counts).value_counts().rename_axis('cell_type').reset_index(name='count')
```

```python
mapping = {'.': 0, '#': 1, 'S': 2, 'G': 3}
numeric = df.replace(mapping).astype(int)

plt.figure(figsize=(5,5))
plt.imshow(numeric.values, interpolation='nearest')
plt.title('Grid Layout Heatmap')
plt.colorbar(ticks=[0,1,2,3], label='Cell Type')
plt.xticks(range(numeric.shape[1]), numeric.columns)
plt.yticks(range(numeric.shape[0]), numeric.index)
plt.show()
```
